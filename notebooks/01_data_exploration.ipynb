# Notebook 1 — Data Exploration

This notebook demonstrates how to load cell detection results and tissue
segmentation masks, and visualises the raw inputs to the pipeline.

> **Note:** Replace the placeholder paths below with your actual data.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# ------------------------------------------------------------------
# Paths — update these to point at your data
# ------------------------------------------------------------------
DETECTION_CSV  = "path/to/cell_detections.csv"   # columns: x, y, label
TISSUE_MASK    = "path/to/tissue_mask.png"        # integer class labels
WSI_THUMBNAIL  = "path/to/thumbnail.png"          # RGB overview image
```

```python
# Load detections
detections = pd.read_csv(DETECTION_CSV)
print(f"Total detections: {len(detections)}")
detections.head()
```

```python
# Visualise detections on thumbnail
thumbnail = np.array(Image.open(WSI_THUMBNAIL).convert("RGB"))
tissue_mask = np.array(Image.open(TISSUE_MASK))

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(thumbnail)
axes[0].scatter(detections["x"], detections["y"], c="red", s=2, alpha=0.5)
axes[0].set_title("Cell Detections")
axes[0].axis("off")

axes[1].imshow(tissue_mask, cmap="tab10", vmin=0, vmax=9)
axes[1].set_title("Tissue Segmentation Mask")
axes[1].axis("off")

plt.tight_layout()
plt.show()
```

```python
# Label distribution
if "label" in detections.columns:
    detections["label"].value_counts().plot(kind="bar", title="Detection Label Distribution")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()
```
